Calcula el total de ventas por categoría de producto.

```python
beauty = df[df['Product Category'] == 'Beauty']
clothing = df[df['Product Category'] == 'Clothing']
electronics = df[df['Product Category'] == 'Electronics']
total_beauty = np.sum(beauty['Total Amount'])
total_clothing = np.sum(clothing['Total Amount'])
total_electronics= np.sum(electronics['Total Amount'])
print(f"Total ventas Beauty: {total_beauty}")
print(f"Total ventas Clothing: {total_clothing}")
print(f"Total ventas Electronics: {total_electronics}")
```

Calcula el promedio de ventas diarias por categoría de producto.

```python
np.mean(beauty['Total Amount'])
np.mean(clothing['Total Amount'])
np.mean(electronics['Total Amount'])
print(f"Promedio ventas Beauty: {np.mean(beauty['Total Amount']):.2f}")
print(f"Promedio ventas clothing: {np.mean(clothing['Total Amount']):.2f}")
print(f"Promedio ventas electronics: {np.mean(electronics['Total Amount']):.2f}")
```

Identifica las categorías de productos con mayores y menores ventas.

```python
totales = np.array([total_beauty, total_clothing, total_electronics])
categorias = np.array(['Beauty', 'Clothing', 'Electronics'])

idx_max = np.argmax(totales)
idx_min = np.argmin(totales)

print(f"Categoría con mas ventas: {categorias[idx_max]} con {totales[idx_max]} ventas")
print(f"Categoría con menos ventas: {categorias[idx_min]} con {totales[idx_min]} ventas")
```

porcentaje del total de ventas por categoria

```python
total_general = np.sum([total_beauty, total_clothing, total_electronics])
print(f"Total ventas general: {total_general}")
porcentaje_beauty = total_beauty / total_general *100
print(f"Beauty: {porcentaje_beauty:.2f}%")
porcentaje_clothing = total_clothing / total_general *100
print(f"clothing: {porcentaje_clothing:.2f}%")
porcentaje_electronics = total_electronics / total_general *100
print(f"electronics: {porcentaje_electronics:.2f}%")
```
